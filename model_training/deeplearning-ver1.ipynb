```python
import h5py
import numpy as np

def load_h5py_file(file_path):
    data = {
        'neural_features': [],
        'n_time_steps': [],
        'seq_class_ids': [],
        'seq_len': [],
        'transcriptions': [],
        'sentence_label': [],
        'session': [],
        'block_num': [],
        'trial_num': [],
    }
    with h5py.File(file_path, 'r') as f:
        keys = list(f.keys())
        for key in keys:
            g = f[key]
            neural_features = g['input_features'][:]
            n_time_steps = g.attrs.get('n_time_steps')
            seq_class_ids = g.get('seq_class_ids')
            seq_class_ids = seq_class_ids[:] if seq_class_ids is not None else None
            seq_len = g.attrs.get('seq_len')
            transcription = g.get('transcription')
            transcription = transcription[:] if transcription is not None else None
            sentence_label = g.attrs.get('sentence_label')
            session = g.attrs.get('session')
            block_num = g.attrs.get('block_num')
            trial_num = g.attrs.get('trial_num')

            data['neural_features'].append(neural_features)
            data['n_time_steps'].append(n_time_steps)
            data['seq_class_ids'].append(seq_class_ids)
            data['seq_len'].append(seq_len)
            data['transcriptions'].append(transcription)
            data['sentence_label'].append(sentence_label)
            data['session'].append(session)
            data['block_num'].append(block_num)
            data['trial_num'].append(trial_num)
    return data


if __name__ == "__main__":
    file_path = "/kaggle/input/brain-to-text-25/t15_copyTask_neuralData/hdf5_data_final/t15.2023.08.13/data_val.hdf5"
    
    data = load_h5py_file(file_path)

    print("\nLoaded file summary:")
    print(f"number of instances in a day: {len(data['neural_features'])}")
    print(f"Neural features shape (first trial): {data['neural_features'][0].shape}")
    print(f"First transcription: {data['transcriptions'][0]}")
    print(f"Sentence label: {data['sentence_label'][0]}")
    print(f"len of one transcription vector: {len(data['transcriptions'][0])}")
```
